In this section, we compare unemployment trends before and during the COVID-19 pandemic, using the available dates and relevant variables to identify notable changes.

```python
Unemployment_all = pd.concat(
    [Unemployment1, Unemployment2],
    ignore_index=True
)

print("Combined shape:", Unemployment_all.shape)
print("Duplicate rows:", Unemployment_all.duplicated().sum())
```

```python
Unemployment_all.columns = Unemployment_all.columns.str.strip()
```

```python
Unemployment_all = Unemployment_all.loc[
    :, ~Unemployment_all.columns.duplicated()
]
```

```python
date_data = Unemployment_all['Date']
if isinstance(date_data, pd.DataFrame):
  date_data = date_data.iloc[:, 0]
```

```python
Unemployment_all['Date'] = pd.to_datetime(
    Unemployment_all['Date'].astype(str), errors='coerce'
)
```

```python
rate_col = [
    c for c in Unemployment_all.columns if 'Unemployment Rate' in c
][0]
state_col = 'Region' if 'Region' in Unemployment_all.columns else 'States'
```

```python
covid_start = pd.to_datetime('2020-03-25')
covid_end = pd.to_datetime('2020-05-31')
```

```python
conditions = [
    (Unemployment_all['Date'] >= covid_start)
    & (Unemployment_all['Date'] <= covid_end),
    (Unemployment_all['Date'] < covid_start),
]
```

```python
choices = [
    'COVID-19 Lockdown (Mar-May 2020)',
    'Pre-COVID (May 2019 - Mar 2020)',
]
```

```python
Unemployment_all['Period'] = np.select(
    conditions, choices, default='Post-Lockdown (June 2020)'
)
```

```python
top_30_unemp = Unemployment_all.nlargest(30, rate_col)
```

```python
top_5_modal_states_counts = (
    top_30_unemp[state_col].value_counts().head(5).reset_index()
)
top_5_modal_states_counts.columns = ['State', 'Frequency']

top_5_modal_dates = (
    top_30_unemp['Date'].dt.strftime('%Y-%m-%d').value_counts().head(5).reset_index()
)
top_5_modal_dates.columns = ['Date', 'Frequency']

print('--- Top 5 Modal States in Top 30 High Unemployment Records ---')
print(top_5_modal_states_counts)

print('\n--- Top 5 Modal Dates in Top 30 High Unemployment Records ---')
print(top_5_modal_dates)
```

# Key Findings and Conclusion

In this final section, we summarize the major findings, highlight important unemployment trends, discuss the observed changes during COVID-19, and acknowledge any limitations of the analysis.

### Rural vs. Urban Unemployment Disparity
The analysis reveals a distinct geographical divide, with urban areas recording a higher average estimated unemployment rate of over 13%, compared to rural areas which stand at approximately 10.3%
This disparity indicates that urban and metropolitan labor markets—often heavily reliant on services, retail, construction, and organized/unorganized daily-wage commerce—faced more acute employment disruptions and structural bottlenecks during the economic shock

### Temporal Trajectory & COVID-19 Lockdown Shock
Pre-COVID Baseline Consistency (May 2019 – March 2020): National unemployment rates remained tightly regulated and stable, hovering predictably between 9% and 10%
Pre-Lockdown Baseline Stability: Prior to early 2020, both rural and urban unemployment rates maintained relatively stable, predictable trajectories, with urban rates consistently tracking slightly higher than rural rates across the months
The Pandemic Shock (March–May 2020): The timeline captures an unprecedented, explosive vertical surge in unemployment coinciding precisely with the nationwide lockdown implementation, causing urban unemployment to skyrocket to a peak near 28% and rural unemployment to spike above 21%
Post-Peak Correction: Following the acute crisis peak between March and May 2020, a steep downward correction occurs heading into mid-2020, highlighting a partial recovery and the resumption of economic activities as restrictions began to ease
Absolute Temporal Concentration: The intersection matrix provides definitive visual proof that 100% of the top 30 highest unemployment rate spikes in the dataset are exclusively clustered within the March 2020 and May 2020 lockdown timeline.
The total absence of high-spike records across the entire pre-COVID timeline (May 2019 through January 2020) confirms that these unprecedented surges were a direct, immediate consequence of nationwide pandemic restrictions rather than prior structural trends.


### Regional Disparities & State-Level Vulnerability
Pronounced Geographic Variance: The regional breakdown reveals a sharp divide, with states like Tripura, Haryana, Jharkhand, and Bihar recording the highest average unemployment rates, peaking near or above 25%.
Vulnerability Hotspots: highest average unemployment rates, peaking near or above 25%
Resilient Lower Tiers: Certain regions display significantly higher susceptibility to labor market contractions compared to others
Policy Implication: These stark disparities emphasize that macro-level interventions must be tailored regionally, as uniform economic policies fail to address the unique structural vulnerabilities of high-risk states during systemic shocks

### Most Vulnerable States in Top 30 Unemployment Peaks
Concentration of Extreme Outliers: The frequency distribution reveals that states like Haryana (leading with 5 appearances), Jharkhand, Bihar, and Tripura (each with 4), alongside Puducherry (with 3), dominate the top 30 highest unemployment rate records in the dataset
Systemic Regional Disruption: The high recurrence of these specific regions within the extreme threshold indicates that their local labor markets experienced repeated, severe shocks during the pandemic restrictions rather than isolated anomalies.
Socio-Economic Factors: States heavily reliant on informal economies, daily-wage labor, and migrant workforces faced disproportionately higher compounding pressures during the nationwide lockdown window.

### State-wise Comparison (Pre-COVID vs. COVID-19 Lockdown)
Widespread Regional Expansion: Across virtually all states, average unemployment rates during the COVID-19 lockdown (red bars) significantly surpassed their pre-COVID baselines (dark blue bars).
Certain regions, such as Puducherry, Jharkhand, Haryana, and Tripura, experienced severe spikes, with average unemployment rates during the lockdown window stretching toward or exceeding 40%.
The comparative grouping demonstrates that the pandemic shock was macro-level and systemic, heavily expanding labor market distress across every state regardless of its regional baseline.


## Strategic Recommendations & Policy Suggestions
### Implement Tailored, Region-Specific Economic Support
Because states like Haryana, Jharkhand, Bihar, Tripura, and Puducherry experienced disproportionately severe spikes (with rates soaring past 40–70% during lockdown peaks), economic recovery packages and fiscal stimuli should not follow a one-size-fits-all national model. High-risk states require localized job-creation funds and targeted financial aid.   
### Protect Urban Informal and Daily-Wage Laborers
With urban unemployment significantly outpacing rural averages and hitting extreme peaks near 28%, future economic shocks will hit service, construction, and daily-wage sectors hardest. Policymakers should build formalized safety nets, direct benefit transfers, and portable welfare schemes for urban informal and migrant workers who lose livelihoods overnight.   
### Adopt Phased, Localized Containment Measures
The data proves that absolute nationwide lockdowns cause catastrophic, synchronized labor market collapses. Future crisis-management protocols should favor targeted, micro-containment zones or sector-specific phased restrictions rather than blanket shutdowns, allowing essential economic engines to keep functioning.  
### Scale Up Rural Shock Absorbers (e.g., Employment Guarantees)
Since millions of workers reverse-migrate from urban centers to rural areas during sudden crises, rural employment schemes (such as MGNREGA in India) must act as immediate shock absorbers. Expanding work allocations and timely wage disbursements in rural sectors can stabilize household incomes during systemic downturns.
